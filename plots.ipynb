```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Omslag
x = np.linspace(0, 2*np.pi, 150)
waves = [lambda x, mult=mult: np.sin(mult * x) for mult in range(1, 4)]

fig, ax = plt.subplots(figsize=(9, 3))
for wave in waves:
    ax.plot(x, wave(x))

# Center the axes: x-axis in the middle, y-axis on the left
ax.spines['bottom'].set_position('zero')
ax.spines['left'].set_position('zero')
ax.spines['top'].set_color('none')
ax.spines['right'].set_color('none')

ax.xaxis.set_ticks_position('bottom')
ax.yaxis.set_ticks_position('left')

# x ticks as fractions of pi
n_ticks = 4
tick_vals = [k * np.pi/2 for k in range(1, n_ticks+1)]
tick_labels = [r'$\pi/2$', r'$\pi$', r'$3\pi/2$', r'$2\pi$']
ax.set_xticks(tick_vals)
ax.set_xticklabels(tick_labels)

ax.set_xlim(0, 2*np.pi)
ax.set_ylim(-1.2, 1.2)

# Arrow tips at the end of each axis
ax.plot(1, 0, ">k", transform=ax.get_yaxis_transform(), clip_on=False, markersize=8)
ax.plot(0, 1, "^k", transform=ax.get_xaxis_transform(), clip_on=False, markersize=8)

ax.grid()

plt.show()
```
